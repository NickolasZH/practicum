## Предсказание температуры стали на комбинате "Стальная птица"

## Описание проекта
Металлургический комбинат «Стальная птица» стремится снизить потребление электроэнергии на этапе обработки стали, контролируя температуру сплава. Для этого необходимо разработать модель, которая будет предсказывать температуру расплавленной стали на основе различных параметров технологического процесса. Модель должна позволить эффективно управлять процессами плавки и легирования стали, что в свою очередь приведет к оптимизации энергозатрат.

**Цель проекта**  
Разработать модель машинного обучения для предсказания температуры расплавленной стали в процессе её обработки. Модель должна учитывать различные факторы, такие как подача сыпучих материалов, газа, проволоки и параметры работы электродов, чтобы точно прогнозировать температуру с минимальной ошибкой.

### Первичная предобработка данных

#### Отрицательная реактивная мощность
Удалим ключи целиком, где встречаются отрицательные значения в столбце "Реактивная мощность"

````python
# Находим все ключи, для которых есть отрицательные значения в столбце 'Реактивная мощность'
invalid_keys = data_arc_new[data_arc_new['Реактивная мощность'] < 0]['key'].unique()

# Удаляем все строки с этими ключами
data_arc_new_cleaned = data_arc_new[~data_arc_new['key'].isin(invalid_keys)]

# Выводим инфо
display(data_arc_new_cleaned.describe())
````

#### Генерация новых признаков
data_arc_new: Добавим время нагрева дугой в секундах и общую мощность

#### Признаки с большим количеством пропусков
Удалим признаки из датасетов признаки, в которых более 95% пропусков, оставшиеся пропуски заполним нулями.

````python
def handle_missing_data(df, threshold=0.95):
    """
    Удаляет столбцы с количеством пропущенных значений больше порога и заполняет остальные нулями.
    
    Parameters:
        df (DataFrame): Входной датафрейм.
        threshold (float): Максимально допустимая доля пропусков в столбце.
    
    Returns:
        DataFrame: Очищенный датафрейм.
    """
    # Вычисляем процент пропусков по столбцам
    missing_percentage = df.isnull().mean()

    # Выбираем только столбцы с долей пропусков ниже порога
    columns_to_keep = missing_percentage[missing_percentage < threshold].index

    # Создаём копию датафрейма с нужными столбцами
    df_cleaned = df[columns_to_keep].copy()

    # Заполняем оставшиеся пропуски нулями
    df_cleaned = df_cleaned.fillna(0)

    return df_cleaned


# Применение функции для обработки пропусков в каждом датасете
data_bulk_new_cleaned = handle_missing_data(data_bulk_new)
data_bulk_time_new_cleaned = handle_missing_data(data_bulk_time_new)
data_wire_new_cleaned = handle_missing_data(data_wire_new)
data_wire_time_new_cleaned = handle_missing_data(data_wire_time_new)

# data_bulk_new_cleaned = data_bulk_new
# data_bulk_time_new_cleaned = data_bulk_time_new
# data_wire_new_cleaned = data_wire_new
# data_wire_time_new_cleaned = data_wire_time_new

# Печать информации о типах данных и количестве пропусков
print("Обработанный data_bulk_new:")
print(data_bulk_new_cleaned.info())
print("\nОбработанный data_bulk_time_new:")
print(data_bulk_time_new_cleaned.info())
print("\nОбработанный data_wire_new:")
print(data_wire_new_cleaned.info())
print("\nОбработанный data_wire_time_new:")
print(data_wire_time_new_cleaned.info())
````

#### Проверка корректности данных подачи сыпучих материалов и проволоки

````python
def check_material_consistency(volume_df, time_df, label=""):
    """
    Проверяет соответствие между объёмом подачи материала и временем подачи.
    
    volume_df — DataFrame с объёмами подачи (например, data_wire_new)
    time_df — DataFrame со временем подачи (например, data_wire_time_new)
    label — строка-название для вывода (например, "Wire")

    Возвращает:
    - количество ошибок (объём есть — времени нет),
    - список ключей с ошибками,
    - датафрейм с примерами ошибок.
    """
    # Список признаков без 'key'
    cols = [col for col in volume_df.columns if col != 'key']

    # Булевы маски: где есть объём и где есть время
    volume_mask = volume_df[cols].notnull()
    time_mask = time_df[cols].notnull()

    # Ошибки: есть объём, но нет времени
    inconsistencies = volume_mask & ~time_mask
    n_inconsistent = inconsistencies.sum().sum()

    print(f"[{label}] Найдено {n_inconsistent} ячеек с объёмом, но без времени подачи.")

    # Ключи с ошибками
    error_keys = volume_df.loc[inconsistencies.any(axis=1), 'key'].unique()
    print(f"[{label}] Уникальных ключей с ошибками: {len(error_keys)}")

    # Примеры строк с ошибками
    error_rows = volume_df[volume_df['key'].isin(error_keys)]

    return n_inconsistent, error_keys, error_rows
````

````python
# Проверка корректности подачи проволоки
errors_wire, wire_keys, wire_rows = check_material_consistency(
    data_wire_new,
    data_wire_time_new,
    label="Wire"
)

# Проверка корректности подачи сыпучих материалов
errors_bulk, bulk_keys, bulk_rows = check_material_consistency(
    data_bulk_new,
    data_bulk_time_new,
    label="Bulk"
)

# Отображение примеров
print("\nПримеры ошибок в подаче проволоки:")
display(wire_rows.head())

print("\nПримеры ошибок в подаче сыпучих материалов:")
display(bulk_rows.head())
````

#### Проверка на нормальность распределения
Проверим нормальность распределения признаков с помощью стат.тестов.

````python
# Функция для проверки нормальности распределения
def check_normality(data, feature_name):
    feature_data = data[feature_name].dropna()

    # Если мало данных — уменьшим выборку, иначе используем до 5000
    sample_size = min(5000, len(feature_data))
    sample_data = np.random.choice(feature_data, size=sample_size, replace=False)

    print(f"\nПризнак: {feature_name}")
    
    # Тест Шапиро-Уилка
    stat_shapiro, p_shapiro = shapiro(sample_data)
    print(f"Shapiro-Wilk test: статистика={stat_shapiro:.8f}, p-значение={p_shapiro:.8f}")
    
    # Тест Д'Агостино и Пирсона
    stat_normaltest, p_normaltest = normaltest(sample_data)
    print(f"D’Agostino and Pearson test: статистика={stat_normaltest:.8f}, p-значение={p_normaltest:.8f}")
    
    # Тест Колмогорова-Смирнова
    stat_kstest, p_kstest = kstest(sample_data, 'norm', args=(sample_data.mean(), sample_data.std()))
    print(f"Kolmogorov-Smirnov test: статистика={stat_kstest:.8f}, p-значение={p_kstest:.8f}")
    
    # Интерпретация
    if p_shapiro < 0.05 or p_normaltest < 0.05 or p_kstest < 0.05:
        print("Распределение статистически значимо отличается от нормального (отклоняем H₀).")
    else:
        print("Распределение статистически не отличается от нормального (не отклоняем H₀).")

# Применяем ко всем числовым признакам
numeric_features = merged_data.select_dtypes(include=['float64']).columns

for feature in numeric_features:
    check_normality(merged_data, feature)
````

## Подготовка и обучение моделей

### Выбор метрики для задачи

Почему подходит **MAE (Mean Absolute Error)**

1. **Характер задачи:**
   - Мы решаем задачу **регрессии**, где целевая переменная `target_temp` — непрерывная температура в °C.
   - Важно оценить **среднюю абсолютную ошибку прогноза модели** в тех же единицах (градусах), что делает метрику MAE интерпретируемой и приближенной к реальному производству.

2. **Физическая интерпретация:**
   - Абсолютная ошибка (например, 5°C) интуитивно понятна инженерам и операторам, в отличие от MSE или RMSE, которые дают ошибку в квадратных градусах или завышают вклад крупных отклонений.

3. **Устойчивость к выбросам:**
   - В отличие от RMSE, MAE **менее чувствительна к выбросам**, что важно в производственной среде, где данные могут содержать редкие экстремумы.

4. **Целевое значение:**
   - Заказчик установил допустимый уровень MAE — **не более 6.8°C** на тестовой выборке. Это позволяет использовать MAE как **ключевую метрику качества** для принятия решений.

---

Почему дополнительно рассчитывается **R² (коэффициент детерминации)**

1. **Объяснённая дисперсия:**
   - R² показывает **долю дисперсии целевой переменной, объясняемую моделью**. Это помогает понять, насколько хорошо модель захватывает закономерности в данных.

2. **Интерпретация:**
   - Значение R² варьируется от 0 до 1 (или может быть отрицательным при плохих моделях):
     - **R² = 1** — идеальное предсказание;
     - **R² = 0** — модель не лучше среднего;
     - **R² < 0** — модель хуже простого среднего значения.
   - Например, R² = 0.85 означает, что **85% колебаний температуры объясняются признаками** модели, а остальные 15% — случайны или неучтены.

3. **Информативность для заказчика:**
   - Заказчику важно понимать не только абсолютную ошибку, но и **объём объяснённой информации** — насколько полезна модель с точки зрения общей картины технологического процесса.

---

### Подготовка выборок

````python
# Определим признаки и целевую переменную
X = merged_data.drop(columns=['target_temp'])  # Все, кроме целевой переменной
y = merged_data['target_temp']                # Целевая переменная

# Разделим на обучающую и тестовую выборки (тестовая — 25%)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, 
    test_size=0.25, 
    random_state=RANDOM_STATE  # для воспроизводимости
)

# Проверка размеров выборок
print(f"Размер обучающей выборки: {X_train.shape}")
print(f"Размер тестовой выборки: {X_test.shape}")
````

#### Исследуем обучающую выборку на наличие выбросов

````python
def plot_feature_target_scatter(X, y, max_cols=3, sample_size=1000):
    """
    Строит scatter-плоты для всех числовых признаков против целевой переменной y.
    
    Parameters:
        X (DataFrame): тренировочные признаки.
        y (Series): целевая переменная.
        max_cols (int): сколько графиков в строке.
        sample_size (int): ограничение по выборке (для ускорения).
    """
    # Объединяем X и y для удобства
    df = X.copy()
    df['target_temp'] = y.values

    numeric_cols = df.select_dtypes(include='number').columns.drop('target_temp')
    n_features = len(numeric_cols)
    n_rows = (n_features + max_cols - 1) // max_cols  # округление вверх

    plt.figure(figsize=(5 * max_cols, 4 * n_rows))

    for i, col in enumerate(numeric_cols, 1):
        plt.subplot(n_rows, max_cols, i)
        sns.scatterplot(data=df.sample(min(sample_size, len(df))), x=col, y='target_temp', alpha=0.5)
        plt.title(f'{col} vs target_temp')
        plt.xlabel(col)
        plt.ylabel('target_temp')

    plt.tight_layout()
    plt.show()

plot_feature_target_scatter(X_train, y_train)
````

**Наблюдение:** при **очень большом времени нагрева (>2500 секунд)** температура **не поднимается выше 1620°C**.
1. **Физически это подозрительно**:
   - Сталь при долгом нагреве должна либо перегреться, либо система должна отключиться.
   - Если температура остаётся «низкой» (≤1620°C), а нагрев экстремально длинный — вероятно:
     - ошибка логгера времени,
     - сбой замера температуры,
     - потеря контакта или холостой нагрев.

2. **Для модели это выбросы**:
   - Они **искажают регрессионную зависимость**.
   - **MAE возрастает**, потому что модель "путается": долгий нагрев, но температура как у средних значений.
   - Такие случаи **влияют на деревья (RF, XGB)** и особенно на **линейные модели** — нарушается монотонность.

Удалим такие выбросы из модели

````python
# Объединяем признаки и целевую переменную для фильтрации
train_df = X_train.copy()
train_df['target_temp'] = y_train.values

# Количество строк до удаления
initial_rows = train_df.shape[0]

# Удаляем строки с выбросами:
# время нагрева дугой > 2000 сек и температура ≤ 1620
condition = (train_df['Время нагрева дугой в секундах'] > 2000) & (train_df['target_temp'] <= 1620)
train_df_cleaned = train_df[~condition]

# Количество удалённых строк
removed_rows = initial_rows - train_df_cleaned.shape[0]
print(f"Удалено выбросов: {removed_rows} строк")

# Разделим обратно на X и y
X_train_cleaned = train_df_cleaned.drop(columns=['target_temp'])
y_train_cleaned = train_df_cleaned['target_temp']



plot_feature_target_scatter(X_train_cleaned, y_train_cleaned)
````

Выводим размеры после удаления выбросов

````python
print(f"Размер обучающей выборки после очистки: {X_train_cleaned.shape[0]} строк, {X_train_cleaned.shape[1]} признаков")
````

#### Пайплайн

Зададим модели, скейлеры и гиперпараметры

````python
# Пайплайн с 3 моделями
pipeline = Pipeline([
    ('scaler', StandardScaler()),  # Масштабирование признаков
    ('model', LinearRegression())  # Модель по умолчанию (Linear Regression)
])

# Гиперпараметры для каждой модели
param_grid = [
    # Линейная регрессия
    {
        'model': [LinearRegression()],
        'scaler': [StandardScaler(), MinMaxScaler(), RobustScaler(), 'passthrough']  # Разные способы масштабирования
    },
    
    # Random Forest
    {
        'model': [RandomForestRegressor(random_state=RANDOM_STATE)],
        'model__n_estimators': range(50, 800),  # Количество деревьев в лесу
        'model__max_depth': range(5, 30),  # Максимальная глубина деревьев
        'model__min_samples_split': range(2, 40),  # Минимальное количество образцов для разделения
        'model__min_samples_leaf': range(1, 10),  # Минимальное количество образцов на листе
        'scaler': [StandardScaler(), MinMaxScaler(), RobustScaler(), 'passthrough']
    },
    
    # XGBoost
    {
        'model': [XGBRegressor(random_state=RANDOM_STATE)],
        'model__n_estimators': range(50, 800),  # Количество бустинг-раундов
        'model__max_depth': range(5, 30),  # Максимальная глубина деревьев
        'model__learning_rate': [0.01, 0.1, 0.2, 0.3],  # Скорость обучения
        'model__subsample': [0.8, 0.9, 1.0],  # Подсэмплинг
        'scaler': [StandardScaler(), MinMaxScaler(), RobustScaler(), 'passthrough']
    }
]
````

#### Поиск лучшей модели

````python
# RandomizedSearchCV
random_search = RandomizedSearchCV(
    pipeline,
    param_distributions=param_grid,
    n_iter=50,
    cv=5,
    scoring='neg_mean_absolute_error',
    n_jobs=-1,
    verbose=1,
    random_state=RANDOM_STATE
)

# Замер времени обучения
start_time = time.time()
random_search.fit(X_train_cleaned, y_train_cleaned)
end_time = time.time()

# Вывод времени
elapsed_time = end_time - start_time
print(f"\nВремя подбора модели: {elapsed_time:.2f} секунд")

# Лучшая модель
best_model = random_search.best_estimator_
print("\nЛучшая модель:")
display(best_model)

# Метрики на тесте
y_pred = best_model.predict(X_test)
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f"\nЛучший MAE на тесте: {mae:.2f}")
print(f"R² на тесте: {r2:.2f}")

# DummyRegressor
dummy = DummyRegressor(strategy="mean")
dummy.fit(X_train, y_train)
dummy_pred = dummy.predict(X_test)

dummy_mae = mean_absolute_error(y_test, dummy_pred)
dummy_r2 = r2_score(y_test, dummy_pred)

print(f"\nDummyRegressor MAE: {dummy_mae:.2f}")
````

Выводим лучший результат на кросс-валидации (средний по 5 фолдам)
