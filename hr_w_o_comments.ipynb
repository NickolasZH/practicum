````python
# Функция для построения гистограмм
def plot_histograms(dataframe, title):
    # Отбираем только числовые столбцы, кроме 'id'
    numeric_columns = dataframe.select_dtypes(include=['float64', 'int64']).drop(columns='id', errors='ignore')
    
    # Построение гистограмм
    numeric_columns.hist(bins=20, figsize=(15, 10), edgecolor='black')
    plt.suptitle(title, fontsize=16)
    plt.show()

# Построение гистограмм для каждой таблицы
# plot_histograms(train_job_satisfaction_rate, "Распределение количественных признаков (train_job_satisfaction_rate)")
# plot_histograms(test_job_satisfaction_features, "Распределение количественных признаков (test_job_satisfaction_features)")
# plot_histograms(test_target_job_satisfaction_rate, "Распределение количественных признаков (test_target_job_satisfaction_rate)")
````

````python
# Функция для построения графиков
def plot_discrete_and_continuous_combined(dataframe, discrete_features, continuous_features, title):
    fig, axes = plt.subplots(
        len(discrete_features) + len(continuous_features) * 2, 1, figsize=(10, 7 * (len(discrete_features) + len(continuous_features)))
    )
    fig.suptitle(title, fontsize=16)
    
    # Построение графиков для дискретных признаков
    for i, feature in enumerate(discrete_features):
        sns.countplot(data=dataframe, y=feature, ax=axes[i])
        axes[i].set_title(f'Распределение: {feature}')
        axes[i].set_xlabel(feature)
        axes[i].set_ylabel('Частота')
    
    # Построение горизонтальных ящиков с усами и гистограмм для непрерывных признаков
    for j, feature in enumerate(continuous_features, start=len(discrete_features)):
        sns.boxplot(data=dataframe, x=feature, ax=axes[j * 2 - len(discrete_features)])
        axes[j * 2 - len(discrete_features)].set_title(f'Распределение: {feature}')
        axes[j * 2 - len(discrete_features)].set_xlabel(feature)
        axes[j * 2 - len(discrete_features)].set_ylabel('')

        sns.histplot(data=dataframe, x=feature, kde=True, bins=20, ax=axes[j * 2 - len(discrete_features) + 1])
        axes[j * 2 - len(discrete_features) + 1].set_title(f'Распределение: {feature}')
        axes[j * 2 - len(discrete_features) + 1].set_xlabel(feature)
        axes[j * 2 - len(discrete_features) + 1].set_ylabel('Частота')

    plt.tight_layout(rect=[0, 0, 1, 0.95])
    plt.show()

# Определяем дискретные и непрерывные признаки
discrete_features = ['employment_years', 'supervisor_evaluation']
continuous_features = ['salary', 'job_satisfaction_rate']

# Построение графиков для тренировочной выборки
plot_discrete_and_continuous_combined(
    train_job_satisfaction_rate,
    discrete_features=discrete_features,
    continuous_features=continuous_features,
    title="Распределение признаков (train_job_satisfaction_rate)"
)

# Построение графиков для тестовой выборки
plot_discrete_and_continuous_combined(
    test_job_satisfaction_features,
    discrete_features=discrete_features,
    continuous_features=['salary'],  # job_satisfaction_rate отсутствует
    title="Распределение признаков (test_job_satisfaction_features)"
)

# Для целевого признака тестовой выборки
sns.boxplot(data=test_target_job_satisfaction_rate, x='job_satisfaction_rate', orient='h')
plt.title("Распределение job_satisfaction_rate (ящик с усами)")
plt.xlabel('job_satisfaction_rate')
plt.ylabel('')
plt.show()

sns.histplot(data=test_target_job_satisfaction_rate, x='job_satisfaction_rate', kde=True, bins=20)
plt.title("Распределение job_satisfaction_rate (гистограмма)")
plt.xlabel('job_satisfaction_rate')
plt.ylabel('Частота')
plt.show()
````

````python
def check_normality(data, feature_name):
    print(f"Проверка нормальности распределения признака: {feature_name}\n")
    
    # Тест Шапиро-Уилка
    stat_shapiro, p_shapiro = shapiro(data)
    print(f"Shapiro-Wilk test: статистика={stat_shapiro:.4f}, p-значение={p_shapiro:.4f}")
    
    # Тест Д'Агостино и Пирсона
    stat_normaltest, p_normaltest = normaltest(data)
    print(f"D’Agostino and Pearson test: статистика={stat_normaltest:.4f}, p-значение={p_normaltest:.4f}")
    
    # Тест Колмогорова-Смирнова
    stat_kstest, p_kstest = kstest(data, 'norm', args=(data.mean(), data.std()))
    print(f"Kolmogorov-Smirnov test: статистика={stat_kstest:.4f}, p-значение={p_kstest:.4f}")
       
    # Интерпретация
    if p_shapiro < 0.05 or p_normaltest < 0.05 or p_kstest < 0.05:
        print("\nРаспределение статистически значимо отличается от нормального (отклоняем H0).\n")
    else:
        print("\nРаспределение статистически не отличается от нормального (не отклоняем H0).\n")

# Проверка нормальности для непрерывных признаков
for feature in ['salary', 'job_satisfaction_rate']:
    if feature in train_job_satisfaction_rate.columns:
        check_normality(train_job_satisfaction_rate[feature], feature)
````

````python
# Пайплайн для обработки Ordinal признаков
ord_pipe = Pipeline([
    ('replace_missing', FunctionTransformer(lambda x: x.replace(' ', np.nan), validate=False)),  # Заменить пробелы на NaN
    ('imputer', SimpleImputer(missing_values=np.nan, strategy='most_frequent')),  # Заполнение пропусков
    ('encoder', OrdinalEncoder(categories=[
        ['junior', 'middle', 'sinior'],  # level
        ['low', 'medium', 'high'],  # workload
    ], handle_unknown='use_encoded_value', unknown_value=np.nan))  # Присваиваем NaN неизвестным категориям
])

# Пайплайн для обработки OneHot признаков
ohe_pipe = Pipeline([
    ('replace_missing', FunctionTransformer(lambda x: x.replace(' ', np.nan), validate=False)),  # Заменить пробелы на NaN
    ('imputer', SimpleImputer(missing_values=np.nan, strategy='most_frequent')),  # Заполнение пропусков наиболее частым значением
    ('encoder', OneHotEncoder(drop='first', handle_unknown='ignore', sparse_output=False))  # OneHotEncoding с игнорированием неизвестных значений
])

# Комбинированный ColumnTransformer для всех типов признаков
preprocessor = ColumnTransformer([
    ('ord', ord_pipe, ord_columns),  # Обработка Ordinal признаков
    ('ohe', ohe_pipe, ohe_columns),  # Обработка OneHot признаков
    ('num', StandardScaler(), num_columns)   # Обработка количественных признаков
],
    remainder='passthrough'
)
````

### RandomizedSearchCV

Зададим модели и параметры для RandomizedSearchCV

````python
# Создание пайплайна для моделей
pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('models', LinearRegression())  # по умолчанию Linear Regression
])
````

````python
# Гиперпараметры для поиска
param_grid = [
    # словарь для модели DecisionTreeRegressor()
    {
        'models': [DecisionTreeRegressor(random_state=RANDOM_STATE)],
        'models__max_depth': range(1, 30),  # Глубина дерева
        'models__min_samples_split': range(2, 30),  # Минимальное количество для разделения
        'models__min_samples_leaf': range(1, 30),  # Минимальное количество в листьях
        'models__max_features': range(1, 30),  # Возможные стратегии выбора признаков
        'preprocessor__num': [StandardScaler(), MinMaxScaler(), RobustScaler(), QuantileTransformer(), Normalizer(), 'passthrough']  # Разные скейлеры
    },

    # словарь для модели LinearRegression()
    {
        'models': [LinearRegression()],
        'preprocessor__num': [StandardScaler(), MinMaxScaler(), RobustScaler(), QuantileTransformer(), Normalizer(), 'passthrough']
    },

    # словарь для модели Ridge (регуляризованный линейный регрессор)
    {
        'models': [Ridge()],
        'models__alpha': [0.01, 0.1, 1, 10, 100],  # Регуляризация
        'preprocessor__num': [StandardScaler(), MinMaxScaler(), RobustScaler(), QuantileTransformer(), Normalizer(), 'passthrough']
    },

    # словарь для модели Lasso (с L1-регуляризацией)
    {
        'models': [Lasso()],
        'models__alpha': [0.01, 0.1, 1, 10, 100, 500],  # Регуляризация
        'preprocessor__num': [StandardScaler(), MinMaxScaler(), RobustScaler(), QuantileTransformer(), Normalizer(), 'passthrough']
    },

    # словарь для модели ElasticNet (комбинированная L1 и L2 регуляризация)
    {
        'models': [ElasticNet()],
        'models__alpha': [0.01, 0.1, 1, 10],  # Регуляризация
        'models__l1_ratio': [0.1, 0.5, 0.9],  # Соотношение L1 и L2 регуляризации
        'preprocessor__num': [StandardScaler(), MinMaxScaler(), RobustScaler(), QuantileTransformer(), Normalizer(), 'passthrough', PowerTransformer()]
    }
]
````

````python
# Уберём лишние предупреждения, не влияющие на результат
# warnings.filterwarnings("ignore", category=FutureWarning)

# Создание RandomizedSearchCV для подбора гиперпараметров
smape_scorer = make_scorer(smape, greater_is_better=False)

# Перебор гиперпараметров с RandomizedSearchCV
randomized_search = RandomizedSearchCV(
    pipeline, 
    param_distributions=param_grid,
    n_iter=100,  # Увеличим количество итераций для более детального поиска
    cv=5,       # Кросс-валидация
    scoring=smape_scorer,  # Оценка модели с помощью SMAPE
    n_jobs=-1,  # Использование всех доступных ядер
    verbose=1,  # Подробный вывод
    random_state=RANDOM_STATE
)

# Замер времени
start_time = time.time()

# Обучение модели с RandomizedSearchCV
randomized_search.fit(X_train_1, y_train_1)

# Время завершения обучения
end_time = time.time()

# Вычисляем время, потраченное на обучение модели
execution_time = end_time - start_time

best_model_1 = randomized_search.best_estimator_
best_metric_1 = randomized_search.best_score_

# Лучшие параметры модели
display('Лучшая модель и её параметры:', best_model_1)
print('Метрика на валидационной выборке:', best_metric_1 * -1)

# Применяем модель для предсказания на обработанных тестовых данных
y_pred = best_model_1.predict(X_test_1)

# Оценка на валидационной выборке с использованием smape
smape_test = smape(y_test_1, y_pred)
print(f'SMAPE на тестовой выборке: {smape_test}')

# Выводим время, затраченное на выполнение пайплайна
print(f'Время, затраченное на выполнение пайплайна: {execution_time:.2f} секунд')
````

### ColumnTransformer

````python
# Пайплайн для Ordinal признаков
ord_pipe_2 = Pipeline([
    ('replace_missing', FunctionTransformer(lambda x: x.replace({' ': np.nan}), validate=False)),  # Замена пробелов на NaN
    ('imputer', SimpleImputer(strategy='most_frequent')),  # Заполнение пропущенных значений
    ('encoder', OrdinalEncoder(categories=[
        ['junior', 'middle', 'sinior'],  # Категории для уровня
        ['low', 'medium', 'high']  # Категории для нагрузки
    ]))
])

# Пайплайн для OneHot признаков
ohe_pipe_2 = Pipeline([
    ('replace_missing', FunctionTransformer(lambda x: x.replace({' ': np.nan}), validate=False)),  # Замена пробелов на NaN
    ('imputer', SimpleImputer(strategy='most_frequent')),  # Заполнение пропущенных значений
    ('encoder', OneHotEncoder(drop='first', handle_unknown='ignore', sparse_output=False))  # Игнорирование неизвестных категорий
])

# Пайплайн для числовых признаков
num_pipe_2 = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),  # Заполнение пропущенных значений медианой
    ('scaler', StandardScaler())  # Масштабирование
])

# Объединённый ColumnTransformer для всех типов признаков
preprocessor_2 = ColumnTransformer([
    ('ord', ord_pipe_2, ord_columns_2),  # Обработка Ordinal признаков
    ('ohe', ohe_pipe_2, ohe_columns_2),  # Обработка OneHot признаков
    ('num', num_pipe_2, num_columns_2)   # Обработка числовых признаков
], remainder='drop')  # Удаление неиспользуемых признаков
````

### RandomizedSearchCV

````python
# Пайплайн для моделей
pipeline_2 = Pipeline([
    ('preprocessor', preprocessor_2),
    ('select_kbest', SelectKBest(score_func=f_classif)),
    ('model', LogisticRegression())
])
````

````python
# Параметры
param_grid_2 = [
    {
        'model': [DecisionTreeClassifier(random_state=RANDOM_STATE)],
        'model__max_depth': range(1, 3),  # Глубина дерева
        'model__min_samples_split': range(2, 4),  # Минимальное количество для разделения
        'model__min_samples_leaf': range(1, 3),  # Минимальное количество в листьях
        'model__max_features': range(1, 3),  # Возможные стратегии выбора признаков
        'preprocessor__num': [StandardScaler(), MinMaxScaler(), 'passthrough']  # Разные скейлеры
    },
    {
        'model': [KNeighborsClassifier()],
        'model__n_neighbors': range(2, 5),
        'preprocessor__num': [StandardScaler(), MinMaxScaler(), 'passthrough']
    },
    {
        'model': [RandomForestClassifier()],
        'model__min_samples_split': range(2, 4),
        'model__min_samples_leaf': range(1, 3),
        'model__max_features': range(1, 3),
        'model__n_estimators': [100, 200, 300],
        'preprocessor__num': [StandardScaler(), MinMaxScaler(), 'passthrough']
    },
    {
        'model': [LogisticRegression()],
        'model__C': [0.1, 1, 100],
        'model__penalty': ['l1'],
        'model__solver': ['liblinear'],
        
        'model__max_iter': [200, 500, 1000],  # Увеличиваем количество итераций
        'preprocessor__num': [StandardScaler(), MinMaxScaler(), 'passthrough']
    }
]
````

````python
# Уберём лишние предупреждения, не влияющие на результат
warnings.filterwarnings("ignore", category=FutureWarning)

# RandomizedSearchCV
search_2 = RandomizedSearchCV(
    pipeline_2,
    param_distributions=param_grid_2,
    n_iter=50,
    scoring='roc_auc',
    cv=5,
    verbose=1,
    n_jobs=-1,
    random_state=RANDOM_STATE
)

# Замер времени
start_time = time.time()

# Обучение модели
search_2.fit(X_train_2, y_train_2)

# Время завершения обучения
end_time = time.time()
execution_time = end_time - start_time

# Лучшая модель
best_model_2 = search_2.best_estimator_
best_params_2 = search_2.best_params_
roc_auc = search_2.best_score_

# Оценка на тестовой выборке
y_pred_proba_2 = best_model_2.predict_proba(X_test_2)[:, 1]
roc_auc_test = roc_auc_score(y_test_2, y_pred_proba_2)

display('Лучшая модель и её параметры:', best_model_2)
print(f'ROC-AUC на кросс-валидации: {roc_auc:.4f}')
print(f'ROC-AUC на тестовой выборке: {roc_auc_test:.4f}')
print(f'Время выполнения: {execution_time:.2f} секунд')
````
